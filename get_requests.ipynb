```python
import requests


def get_request(username):
  r = requests.get(f'https://api.github.com/users/{username}')
  if r.status_code == 200:
      return r.json()
  else:
      print(f"Request failed with status {r.status_code}")
      return None

print(get_request("asdkjhaskjdh12345"))
print(get_request("tumikef"))
```
